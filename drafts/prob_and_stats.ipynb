## Basic Concepts

> This section deserves better content. For now, only a short review of probability theory and the connection with the statistics

The **probability** $\mathbb{P}[\{\dots\}]$ is a measure of the set that maps an event to a real number $[0, 1]$. Instead of dealing with events $E$ (discrete or continuous), event spaces $\mathcal{F}$ and sample spaces $\Omega$, it's easier to transform those events into numbers. Therefore we define a **random variable** $X: \Omega \to \mathbb{R}$, which is just a _map_, to simplify the analysis.

Consider $\Omega = \{$♠️, ❤️, ♣️, ♦️$\}$ and the probability $\mathbb{P}[\{\text{drawing a spades in a standard deck with 13 cards of each suit}\}] = \frac{\text{\# of spades}}{\text{\# of cards}} = \frac{13}{52} = 0.25$. We can map $X($♠️$) = 1, X($❤️$) = 2, X($♣️$) = 3, X($♦️$) = 4$, and the question becomes $\mathbb{P}[X = 1] = \dots = 0.25$.

> Besides $X$ being a function, we call it a _random variable_ because it operates in a non-deterministic domain (therefore _random_) and has multiple possible states/outcomes (therefore _variable_).

> When writing $\mathbb{P}[X = x]$ what we are actually accounting is $\mathbb{P}[\{ \omega \in \Omega : X(\omega)=x \}] = \mathbb{P}[X^{-1}(\{x\})]$.

Whenever we talk about probability distributions, it is important to distinguish between the _ideal_ and the _empirical_. The former is described by the probability function that governs the problem, while the latter is inferred based on observations (collected data). For any measure $m$ taken from the ideal distribution $f$ we can also compute the empirical measure $\hat{m}$ upon $N$ samples, and as $N \to \infty$ we expect $\hat{m} \to m$.

First we are going to delve into ideal probability distributions and their properties, discovering what they can tell about the data. Then we inverse the logic, studying how one can make statements and infer probability properties by looking at the data.

## Discrete Random Variables

### Discrete Distributions

If $X$ is discrete, we define the **probability mass function** (PMF) over $X$ as the probability of obtaining a number $x$,
$$f_X(x) = \mathbb{P}[X = x],$$
and the **cumulative distribution function** (CDF) over $X$ as
$$ F_X(x) =  \mathbb{P}[X \le x] = \sum_{x' \in X(\Omega), x' \le x} f_X(x').$$

Note that, as we satisfy the normalization constraint $\sum_{x \in X(\Omega)} f_X(x) = 1$, the PMF is basically a normalized ideal histogram over $\Omega$. In reality, what we measure is the _empirical_ histogram, and we expect their distributions to converge for large samples.

The **expectation** of $X$ is a _localization_ measure of a random variable, and is given by the average of $x \in X$ weighted by its probability $f_X(x)$, computed from the ideal distribution.

$$\mathbb{E}[X] = \sum_{x \in X(\Omega)} x f_X(x) = \mu$$

> Although it is not guaranteed $\mathbb{E}[X] \in X(\Omega)$, we can think of it as the center of mass of $f_X(x)$.

> Not every PMF has an expectation. For the existence of $\mathbb{E}[X]$ we need it to be _absolute summable_, that is: $\mathbb{E}[|X|] < \infty$.

Also, we can generalize the definition to the $k$-th **moment** of $X$, given by
$$\mathbb{E}[X^k] = \sum_{x \in X(\Omega)} x^k f_X(x).$$

The **variance** of $X$ is a _dispersion_ measure of a random variable, and is, roughly speaking, the expected spread of $x \in X$ around its expected value, computed from the ideal distribution.

$$\text{Var}[X] = \mathbb{E}[(X - \mu)^2] = \sigma^2$$

> It is basically the second moment of the centralized random variable.

Also, it is convenient to define a dispersion measure in the same scale, so we have the **standard deviation**
$$\text{Std}[X] = \sqrt{\text{Var}[X]} = \sigma.$$


For any function $g(x), h(x)$ and any scalar $c$, we have the following properties.
- **function**:
	- $\mathbb{E}[g(X)] = \sum_{x \in X(\Omega)} g(x) f_X(x)$
- **additive**:
	- $\mathbb{E}[g(X) + h(X)] = \mathbb{E}[g(X)] + \mathbb{E}[h(X)]$
- **scale**:
	- $\mathbb{E}[cX] = c\mathbb{E}[X]$
	- $\text{Var}[cX] = c^2\text{Var}[X]$
- **shift**:
	- $\mathbb{E}[X + c] = \mathbb{E}[X] + c$
	- $\text{Var}[X + c] = \text{Var}[X]$
- **moment**:
	- $\text{Var}[X] = \mathbb{E}[X^2] - \mathbb{E}[X]^2$

Some common discrete random variables:

| Name | Use Case | PMF Definition | Domain | Expectation | Variance |
|------|----------|----------------|--------|-------------|----------|
| Bernoulli | binary-state events | $X \sim \text{Bernoulli}(p): \begin{cases} f_X(0) = 1-p \\ f_X(1) = p \end{cases}$ | $x \in \{0, 1\}$ | $\mathbb{E}[X] = p$ | $\text{Var}[X] = p(1-p)$ |
| Binomial  | $x$ successes in $n$ trials | $X \sim \text{Binom}(n, p): f_X(x) = \binom{n}{x} p^x (1-p)^{n-x}$ | $x \in \{0, 1, \dots, n\}$ | $\mathbb{E}[X] = np$ | $\text{Var}[X] = np(1-p)$ |
| Geometric | success at the $x$-th trial | $X \sim \text{Geom}(p): f_X(x) = (1-p)^{x-1} p$ | $x \in \{1, 2, 3, \dots\}$ | $\mathbb{E}[X] = 1/p$ | $\text{Var}[X] = (1-p)/p^2$ |
| Poisson   | events with arrival rate $\lambda$ | $X \sim \text{Poisson}(\lambda): f_X(x) = \frac{\lambda^x}{x!} e^{-\lambda}$ | $x \in \{0, 1, 2, \dots\}$ | $\mathbb{E}[X] = \lambda$ | $\text{Var}[X] = \lambda$ |

## Continuous Random Variables

### Continuous Distributions

In a similar way, if $X$ is continuous, we define the **cumulative distribution function** (CDF) over $X$ as
$$ F_X(x) = \mathbb{P}[X \le x]$$
and, if the CDF is differentiable, the **probability density function** (PDF) over $X$ as the non-negative function
$$ f_X(x) = \frac{d}{dx}F_X(x)$$
such that 
$$ F_X(x) = \int_{-\infty}^x f_X(u) du.$$

> The CDF must always be right-continuous.

That way, the probability of obtaining a number $x \in [a,b]$ is given by
$$\mathbb{P}[a \le X \le b] = \int_a^b f_X(u) du = F_X(b) - F_X(a).$$

> We call $f_X(x)$ a _density_ because it is the probability per unit length, and we need to integrate it since the measure of a unit point is always zero for continuous random variables.

> For the same reason, we can assume $\mathbb{P}[a \le X \le b] = \mathbb{P}[a < X \le b] = \mathbb{P}[a \le X < b] = \mathbb{P}[a < X < b]$.

We can define the **expectation** of $X$ in a similar way, and the definitions of variance and standard deviation remain the same.

$$\mathbb{E}[X] = \int_{X(\Omega)} x f_X(x) dx = \mu$$

Some common continuous random variables:

| Name | Use Case | PDF Definition | Domain | Expectation | Variance |
|------|----------|----------------|--------|-------------|----------|
| Uniform | equally likely events inside an interval | $X \sim \text{Unif}(a, b): \begin{cases} f_X(x) = \frac{1}{b-a} \;\text{if}\; a \le x \le b\\ f_X(x) = 0 \;\text{otherwise} \end{cases}$ | $x \in \mathbb{R}$ | $\mathbb{E}[X] = \frac{a+b}{2}$ | $\text{Var}[X] = \frac{(b-a)^2}{12}$ |
| Exponential | events with decaying rate $\lambda > 0$ | $X \sim \text{Exp}(\lambda): \begin{cases} f_X(x) = \lambda e^{-\lambda x} \;\text{if}\; x \ge 0\\ f_X(x) = 0 \;\text{otherwise} \end{cases}$ | $x \in \mathbb{R}$ | $\mathbb{E}[X] = \frac{1}{\lambda}$ | $\text{Var}[X] = \frac{1}{\lambda^2}$ |
| Gaussian | events _normally_ distributed | $X \sim \mathcal{N}(\mu, \sigma^2): f_X(x) = \frac{1}{\sqrt{2\pi\sigma^2}} \exp\left(- \frac{(x-\mu)^2}{2 \sigma^2}\right)$ | $x \in \mathbb{R}$ | $\mathbb{E}[X] = \mu$ | $\text{Var}[X] = \sigma^2$ |

The Gaussian or Normal distribution is so important that we have a special notation for the CDF of $\mathcal{N}(0, 1)$,
$$\Phi(x) = \frac{1}{\sqrt{2\pi}} \int_{-\infty}^x e^{-u^2/2} du$$
such that for $X \sim \mathcal{N}(\mu, \sigma^2)$ we have
$$F_X(x) = \Phi\left( \frac{x - \mu}{\sigma} \right).$$

> This importance is related to the error function, that will be explained after.

### Higher Order Moments

Moreover, now we can define other central moments called **skewness**, which is a measure of asymmetry (negative if skewed towards left, positive if towards right),
$$\text{Skew}[X] = \mathbb{E}\!\left[\left( \frac{X - \mu}{\sigma} \right)^3 \right] = \gamma,$$
and **kurtosis**, which measures how heavy-tailed is the probability distribution,
$$\text{Kurt}[X] = \mathbb{E}\!\left[\left( \frac{X - \mu}{\sigma} \right)^4 \right] = \kappa.$$

The interpretation of the kurtosis is easier by considering the _excess kurtosis_ $\kappa_\text{excess} = \kappa - 3$ that discounts the value of the Gaussian distribution. If $\kappa_\text{excess} > 0$ then the distribution decays slower than the Gaussian and have higher probability of extreme outliers, and if negative it decays faster, showing less outliers.

## Joint Distributions

### High-Dimensional Distributions

In short, a **joint distribution** is a high-dimensional probability distribution (PMF, PDF, CDF). For one random variable we write $f_{X}(x)$, for 2 variables $f_{X,Y}(x,y)$, for three variables $f_{X,Y,Z}(x,y,z)$, and so on. Thus, it is convenient to use vector notation,
$$f_\mathbf{X}(\mathbf{x}) = f_{X_1, \dots,X_N}(x_1, \dots, x_N),$$
where $\mathbf{X} = [X_1, \dots, X_N]^\top$ is a vector of $N$ random variables and $\mathbf{x} = [x_1, \dots, x_N]^\top$ is a vector of $N$ random states.

Let's start with the simplest case: 2 random variables. Consider the cartesian product of two sample spaces $\Omega_X \times \Omega_Y = \{(\omega_x, \omega_y)\}$, such that $\mathbf{X}: (\omega_x, \omega_y) \to (x, y)$ where $(x, y) = \mathbf{x} \in \mathcal{S}_X \times \mathcal{S}_Y$.

For discrete random variables, the **joint PMF** of $X$ and $Y$ is defined as $\mathbb{P}[X=x, Y=y] = f_{X,Y}(x,y)$, giving
$$\mathbb{P}[\mathcal{A}] = \sum_{(x,y) \in \mathcal{A}} f_{X,Y}(x,y).$$

Similarly, for continuous random variable, we can define the **joint PDF** of $X$ and $Y$ such that
$$\mathbb{P}[\mathcal{A}] = \int_\mathcal{A} f_{X,Y}(x,y) \,dx \,dy.$$

### Marginal Distributions

The **marginal PDF/PMF** of a random variable is the result of summing/integrating every other variable except that one.

For a discrete random variable:
$$ f_X(x) = \sum_{y \in \mathcal{S}_Y} f_{X,Y}(x,y) \quad\text{and}\quad f_Y(y) = \sum_{x \in \mathcal{S}_X} f_{X,Y}(x,y) .$$

For a continuous random variable:
$$ f_X(x) = \int_{\mathcal{S}_Y} f_{X,Y}(x,y)dy \quad\text{and}\quad f_Y(y) = \int_{\mathcal{S}_X} f_{X,Y}(x,y)dx .$$

We call those random variables **independent** if $f_{X,Y}(x,y) = f_{X}(x) f_{Y}(y)$, or for a $N$-dimensional distribution, $f_\mathbf{X}(\mathbf{x}) = \prod_{n=1}^N f_{X_n}(x_n)$. Also, if  they all have the same distribution $f_{X_1}(x_1) = f_{X_2}(x_2) = \dots = f_{X_N}(x_N)$ we say they are **identically distributed**. In the (so desired) case where they are both **independent and identically distributed** (i.i.d.) the joint distribution of an $N$-dimensional random variable is just the $N$-th power of $f_{X_n}(x_n)$, where the choice of $i$ is irrelevant (all give the same result).

For the **joint CDF**, it is simply given by
$$ F_{X,Y}(x,y) = \sum_{x' \le x}\sum_{y' \le y} f_{X,Y}(x',y')$$
for discrete random variables and 
$$ F_{X,Y}(x,y) = \int_{-\infty}^x\int_{-\infty}^y f_{X,Y}(x',y') \,dy'\,dx'$$
for continuous random variables. Note that if $X,Y$ are independent we can split the summation/integration into a product of the marginal PMFs/PDFs.

### Joint Conditionals

Finally, having introduced the joint and marginal distributions, we can define the **joint conditional** PMF/PDF as
$$f_{X|Y}(x|y) = \frac{f_{X,Y}(x,y)}{f_Y(y)}$$
that can be seen as a function of $X$, as the uncertainty of other variable is removed by fixing $Y = y$. For discrete variables, is exactly
$$f_{X|Y}(x|y) = \frac{\mathbb{P}[X=x \cap Y=y]}{\mathbb{P}[Y=y]} = \mathbb{P}[X=x | Y=y],$$
while for continuous variables we must resort to the CDF with the limiting perspective,
$$
\begin{split}
F_{X|Y}(x|y) &= \lim_{h \to 0} \mathbb{P}[X \le x \;|\; y \le Y \le y+h] \\
	&= \lim_{h \to 0} \frac{\mathbb{P}[X \le x \cap y \le Y \le y+h]}{\mathbb{P}[y \le Y \le y+h]} \\
	&= \frac{ \int_{-\infty}^x f_{X,Y}(x',y) \,dx' }{ f_{Y}(y) }
\end{split}
$$
what leads to
$$f_{X|Y}(x|y) = \frac{d}{dx}F_{X|Y}(x|y).$$

### Statistics of Joint Distributions

Regarding the expectation of two random variables $X,Y$, we define the **joint expectation** (also called **correlation**) and the **conditional expectation** as
$$\mathbb{E}[XY] = \sum_{x \in \mathcal{S}_X}\sum_{y \in \mathcal{S}_Y} xy \, f_{X,Y}(x,y)$$
$$\mathbb{E}[X|Y] = \sum_{x \in \mathcal{S}_X} x f_{X|Y}(x|y)$$
in the discrete case and
$$\mathbb{E}[XY] = \int_{\mathcal{S}_X}\int_{\mathcal{S}_Y} xy \, f_{X,Y}(x,y)\,dx\,dy$$
$$\mathbb{E}[X|Y] = \int_{\mathcal{S}_X} x f_{X|Y}(x|y)\,dx$$
in the continuous one. Again, if the variables are independent, we can split $\mathbb{E}[XY] = \mathbb{E}[X]\mathbb{E}[Y]$.

Also, the equivalent of the variance for $X,Y$ is the **covariance**
$$\text{Cov}(X,Y) = \mathbb{E}[(X - \mu_X)(Y - \mu_Y)].$$

Some important properties, for any $X, Y$:
- $\mathbb{E}[X + Y] = \mathbb{E}[X] + \mathbb{E}[Y]$
- $\text{Cov}(X,Y) = \mathbb{E}[XY] - \mathbb{E}[X]\mathbb{E}[Y]$
- $\text{Cov}(X,X) = \text{Var}[X]$
- $\text{Var}[X + Y] = \text{Var}[X] + 2\text{Cov}(X,Y) + \text{Var}[Y]$

Finally, we can define the (Pearson) **correlation coefficient** $\rho \in [-1, 1]$ as the cosine angle between the centralized variables,
$$
\rho = 
\frac{\text{Cov}(X,Y)}{\sqrt{\text{Var}[X]\text{Var}[Y]}} = 
\frac{\mathbb{E}[XY] - \mu_X\mu_Y}{\sigma_X\sigma_Y}.
$$

It is worth to mention the _law of total expectation_, which is intuitively as the expectation of a random variable regardless of the other ones. Think of it as a projection into the axis of the desired variable. Considering $X, Y$ we have
$$
\mathbb{E}[X] 
= \mathbb{E}_Y \!\left[ \mathbb{E}_{X|Y}[X|Y] \right] 
= \begin{cases} 
	\sum_{y \in \mathcal{S}_Y} \mathbb{E}[X|Y] f_Y(y) &\text{for discrete random variables,} \\ 
	\int_{\mathcal{S}_Y} \mathbb{E}[X|Y] f_Y(y)\,dy &\text{for continuous random variables.} 
\end{cases}
$$

For multidimensional random vector (i.e. vectors of random variables) we define the expectation by
$$
\mathbb{E}[\mathbf{X}] 
= \begin{bmatrix} 
	\mathbb{E}[X_1]
	\\ \vdots \\ 
	\mathbb{E}[X_N] 
\end{bmatrix}
= \begin{bmatrix} 
	\int_\mathcal{S} x_1 f_{X_1}(x_1) \,dx_1
	\\ \vdots \\ 
	\int_\mathcal{S} x_N f_{X_N}(x_N)  \,dx_N
\end{bmatrix}
= \mathbf{\mu}
$$
where the marginal PDF is determined by 
$$ f_{X_n}(x_n) = \int_\mathcal{S} f_\mathbf{X}(\mathbf{x}) \,d\mathbf{x}_{\backslash n} \,.$$

We can define the **covariance matrix** of the $N$-dimensional random vector $\mathbf{X}$ as the symmetric matrix
$$
\mathbf{\Sigma}
= \text{Cov}(\mathbf{X})
= \begin{bmatrix} 
	\text{Var}[X_1]      & \text{Cov}(X_1, X_2) & \dots  & \text{Cov}(X_1, X_N) \\
	\text{Cov}(X_2, X_1) & \text{Var}[X_2]      & \dots  & \text{Cov}(X_2, X_N) \\
	\vdots               & \vdots				& \ddots & \vdots				\\
	\text{Cov}(X_N, X_1) & \text{Cov}(X_N, X_2) & \dots  & \text{Var}[X_N]
\end{bmatrix}
= \mathbb{E}\!\left[ (\mathbf{X} - \mathbf{\mu})(\mathbf{X} - \mathbf{\mu})^\top \right].
$$

## Moment-Generating Functions

First, let's consider the following problem: we have two continuous random variables $X, Y$ and we want to determine the sum $Z = X+Y$. It is surprising that $f_Z(z) \neq f_X(x) + f_Y(y)$. In fact, the resulting PDF is given by the _convolution_ of the PDFs
$$ f_Z(z) = (f_X * f_Y)(z) = \int_{-\infty}^{+\infty} f_X(z-y) \, f_Y(y) \,dy .$$
Intuitively, we can reason that every realization of one random variable can combine with all realizations of the other, giving the multiplication that characterizes the convolution.

An easier way of computing this sum is by using moment-generating functions, allowing us to perform the multiplication in a transformed domain instead of the convolution.

A **moment-generating function** (MGF) of a random variable $X$ is basically the Laplace transform of the random variable, which is given by
$$
M_X(s) = \mathbb{E}[e^{sX}]
= \begin{cases}
\sum_{x \in \mathcal{S}_X} e^{sx} f_X(x) &\text{for discrete variables},
\\
\\
\int_{-\infty}^{+\infty} e^{sx} f_X(x)\,dx &\text{for continuous variables}.
\end{cases}
$$

Some special properties of MGFs, where the second is responsible for its name, are
- $M_X(0) = 1$,
- $\frac{d^k}{ds^k} \,M_X(s)|_{s=0} = \mathbb{E}[X^k]$, for any $k \in \mathbb{N}^+$.

Now we can approach the sum of _independent_ random variables $Z = X + Y$ by another perspective, using the following theorem.
$$M_Z(s) = M_X(s) M_Y(s)$$
If we take $Z = \sum_{n=1}^N X_n$ we get the generalization $M_Z(s) = \prod_{n=1}^N M_{X_n}(s)$, and if the random variables are i.i.d. we simplify it by $M_Z(s) = (M_{X_1}(s))^N$.

> As the former integral does not converge for every PDF, not every random variable has a MGF. A way to circumvent it is to restrict $s = -j\omega$ in order to satisfy $|e^{sx}| = 1$, transforming the MGF into a **characteristic function** $\Phi_X(j\omega) = \mathbb{E}[e^{-j\omega X}]$.

> The former definition, with the negative exponent, is intentional to make it compatible with the Fourier transform definition.

## Sample Statistics

Consider the example where we have $N$ observations that can interact with each other. If $N$ is big enough (e.g. in the order of millions), computing the joint probability is computationally unfeasible. A better option is to summarize those observations in sample statistics. Or in other words, transform microstate descriptions into macrostate descriptions.

The first measure is the **sample average** over $N$ random observations
$$ \overline{X}_N = \frac{1}{N} \sum_{n=1}^N X_n.$$
Note that we are using $X_n$ and not $x_n$, as we are defining an **estimator**, that is, a representation before the data collection. It means that every time we run another sampling with $N$ observations, the value of $\overline{X}_N$ can be different. In fact, $\overline{X}_N$ is by itself a random variable, granted its own PDF $f_{\,\overline{X}_N}(x)$ and thus a mean $\mu_{\,\overline{X}_N}$, a variance $\sigma^2_{\,\overline{X}_N}$, and so on.

### Convergence of Sample Statistics

If we take $N$ i.i.d. random variables $X_n$ with (same) population mean $\mu$ and _finite_ variance $\sigma^2$, it is easy to show that
$$ \mathbb{E}\!\left[ \,\overline{X}_N \right] = \frac{1}{N} \sum_{n=1}^N \mathbb{E}[X_n] = \mu$$
and
$$ \text{Var}\!\left[ \,\overline{X}_N \right] = \frac{1}{N^2} \sum_{n=1}^N \text{Var}[X_n] = \frac{\sigma^2}{N},$$
meaning that while the sample expectation is the same as the population expectation, the deviation of $\overline{X}_N$ approaches zero as $N$ grows.

The **"weak" law of large numbers** states that for $N$ i.i.d. random variables $X_n$ with (same) population mean $\mu$ and finite variance $\sigma^2$, for any $\varepsilon > 0$
$$\lim_{N \to \infty} \mathbb{P}\Bigl[ |\overline{X}_N - \mu| > \varepsilon \Bigr] = 0.$$
More importantly, by the Chebyshev's inequality we can estimate how fast the deviation vanishes, given that
$$\mathbb{P}\Bigl[ |\overline{X}_N - \mu| > \varepsilon \Bigr] \le \frac{\sigma^2}{N\varepsilon^2}.$$

This is the so called _convergence by probability_, which means that the sequence of probabilities for growing $N$ must converges to zero in order to assume that it is very unlikely (but not impossible) to observe $|\overline{X}_N - \mu| > \varepsilon$.

By assuming the same statements above as well $\mathbb{E}[X_n^4] < \infty$, the **"strong" law of large numbers** ensures
$$\mathbb{P}\Bigl[ \,\lim_{N \to \infty} \overline{X}_N = \mu \Bigr] = 1,$$
although the intuition behind this proof is not trivial to be summarized here.

A final yet important concept to mention is the **convergence by distribution**, that basically means that the values of the CDF converge for growing $N$.

With more rigour, let's consider $Z_1, Z_2, \dots, Z_N$ random variables (you can imagine discrete ones for the sake of illustration) with CDFs $F_{Z_1}, F_{Z_2}, \dots, F_{Z_N}$. We say this sequence _converges in distribution_ to a random variable $Z$ (for illustration purposes, a continuous one) if
$$\lim_{N \to \infty} F_{Z_N}(z) = F_Z(z).$$

It _does not_ say that every point converges in probability (in the proposed example, the countable points in the discrete PMF $f_{Z_N}$ would never approximate the continuous envelope of the PDF $f_Z$ whose pointwise probability is zero). What it _does_ say is that the pointwise values _in the CDF_ converge.

### Central Limit Theory

Finally, considering $X_1, X_2, \dots, X_N$ i.i.d. random variables with population mean $\mu$, finite variance $\sigma^2 > 0$ and sample average $\overline{X}_N$, and let $Z_N = \sqrt{N} \left(\frac{\overline{X}_N - \mu}{\sigma} \right)$. The **central limit theorem** (CLT) states that the rescaled and centralized sample average, in the form of $Z_N$, converges in distribution to the Gaussian $\mathcal{N}(0,1)$.

Setting aside the proof of this theorem, there is a strong implication of the CLT. In short, it states that the average of any i.i.d. random variables with finite mean and variance will be approximately a Gaussian distribution. It allows us, for example, to ignore the shape of the unknown population distribution, given that as long as we are dealing with sample averages with large $N$ the resulting distribution will be normal.

It is important to clarify that the uncentered and non-rescaled $\overline{X}_N$ is approximated by $\mathcal{N}(\mu, \frac{\sigma^2}{N})$ for large $N$, so its variance vanishes as $N \to \infty$ (converging in probability to the constant $\mu$).

## Parameter Estimation

Now let's invert the logic of the study and estimate the parameters that govern a given population. If we consider that the observed samples are drawn from an unknown distribution $f_X(x; \mathbf{\theta})$, where $\mathbf{\theta}$ is the vector containing one or more model parameters (like the probability $p$ of Bernoulli distribution or the mean $\mu$ and variance $\sigma^2$ of Gaussian distribution).

The task is therefore to recover the unknown parameters based only on the observations, and for that we'll study some useful methods.

### Maximum-Likelihood Estimation

Let's start with the **maximum-likelihood** (ML) **estimation**. In short, the ML consists in finding the parameter $\mathbf{\theta}$ that maximizes the _likelihood function_, given that the observations $x_1, \dots, x_N$ are provided beforehand and therefore are fixed. By maximizing this function, we should discover the $\mathbf{\theta}$ that _better_ describes our data.

Consider a some observations $\{x_1, \dots, x_N \}$. The most general PDF that we can propose is $f_\mathbf{X}(\mathbf{x}; \mathbf{\theta})$, where each $x_n$ is said to be drawn to a specific distribution $X_n$. Although this is descriptive, it can also be too cumbersome (not to mention the computational effort required to calculate the $N \times N$ covariance matrix). If we assume that they are i.i.d. (bold, but very common actually), the PDF becomes $\prod_n f_{X_n}(x_n; \mathbf{\theta})$ and $\mathbf{\Sigma} \to \sigma^2\mathbf{I}_N$, which is much simpler to handle.

Then we define the **likelihood function** as follows.
$$\mathcal{L}(\mathbf{\theta}|\mathbf{x}) \stackrel{\text{def}}{=} \prod_{n = 1}^N f_{X_n}(x_n; \mathbf{\theta})$$
It is convenient to use the **log-likelihood** function in computation as it is more stable numerically.
$$\log \mathcal{L}(\mathbf{\theta}|\mathbf{x}) = \sum_{n = 1}^N \log f_{X_n}(x_n; \mathbf{\theta})$$

> Besides the notation, $\mathcal{L}(\mathbf{\theta}|\mathbf{x})$ is _not_ a conditional PDF as $\mathbf{\theta}$ is not a random variable, but a function of $\mathbf{\theta}$ that changes its values depending on $\mathbf{x}$.

Finally, the _estimate_ parameter $\hat{\mathbf{\theta}}_\text{ML}$ is defined as the parameter $\mathbf{\theta}$ that maximizes $\mathcal{L}$ given $\mathbf{x}$, which is solved by taking the first derivative with respect to $\mathbf{\theta}$ and setting it to zero.
$$
\hat{\mathbf{\theta}}_\text{ML}(\mathbf{x}) \stackrel{\text{def}}{=} \argmax_{\mathbf{\theta}} \mathcal{L}(\mathbf{\theta}|\mathbf{x})
\Rightarrow
\frac{\partial}{\partial\mathbf{\theta}} \Bigl. \mathcal{L}(\mathbf{\theta}|\mathbf{x}) \Bigr|_{\hat{\mathbf{\theta}}_\text{ML}} = \mathbf{0}
$$

> Note that the solution $\hat{\mathbf{\theta}}_\text{ML}$ is the same for the likelihood and the log-likelihood functions, so in practice we solve the second.

> Sometimes can be necessary to also take the second derivative in order to check if the stationary point is a maximum or a minimum (for any concave function, it is not needed).

### Properties of ML Estimators

While $\hat{\mathbf{\theta}}(x_1, \dots, x_N)$ is an estimate given some realizations $x_n$, we can also define the **estimator** $\hat{\mathbf{\Theta}}(X_1, \dots, X_N)$ that takes into account the random variables $X_n$. Note that the former is a number and the latter is a random variable.

One can define a lot of estimators, some will be good, some will be bad. We call an estimator **unbiased** if $\mathbb{E}[\hat{\Theta}] = \theta$, meaning that regardless of the random fluctuations the average estimate matches the true parameter. A nice example is the ML estimator of the Gaussian PDF parameters for some i.i.d. random variables $X_1, \dots, X_N$: while the estimate of the mean converges to the true mean, the estimate of the variance does not (therefore it is biased).

$$
\hat{\mu}_\text{ML} = \frac{1}{N} \sum_{n=1}^N X_n 
\Rightarrow 
\mathbb{E}[\hat{\mu}_\text{ML}] = \mu
$$

$$
\hat{\sigma}^2_\text{ML} = \frac{1}{N} \sum_{n=1}^N (X_n - \hat{\mu}_\text{ML})^2 
\Rightarrow 
\mathbb{E}[\hat{\sigma}^2_\text{ML}] = \frac{N-1}{N}\sigma^2
$$

A possible solution is to define another estimator such that $\hat{\sigma}^2_\text{unbiased} = \frac{1}{N-1} \sum_{n=1}^N (X_n - \hat{\mu}_\text{ML})^2$, even though it does not maximize the likelihood.

We define a **consistent estimator** $\hat{\mathbf{\Theta}}_N = \hat{\mathbf{\Theta}}(X_1, \dots, X_N)$ a function of $N$ variables that converges in probability to the true parameter.

$$\lim_{N \to \infty} \mathbb{P}\left[ \left| \hat{\mathbf{\Theta}}_N - \theta \right| > \varepsilon \right] = 0$$

> Note that consistency does not imply unbiasedness, and vice versa.

### Maximum a Posteriori Estimation

Another interesting estimator is the **maximum a posteriori** (MAP) **estimator**, that leverages a priori knowledge with the observed data. For that, we no more assume the true parameter $\mathbf{\theta}$ to be deterministic and unknown, but a random variable $\mathbf{\Theta}$ with a priori (i.e. provided) PDF $f_\mathbf{\Theta}(\mathbf{\theta})$ with mean $\mu_0$ and variance $\sigma_0^2$.

Considering $f_{\mathbf{X}|\mathbf{\Theta}}(\mathbf{x}|\mathbf{\theta})$ as the likelihood of the MAP and using the Bayes' theorem, the a posteriori PDF is given by
$$
f_{\mathbf{\Theta}|\mathbf{X}}(\mathbf{\theta}|\mathbf{x})
= \frac{
	f_{\mathbf{X}|\mathbf{\Theta}}(\mathbf{x}|\mathbf{\theta}) 
	f_\mathbf{\Theta}(\mathbf{\theta})
}{
	f_\mathbf{X}(\mathbf{x})
}
$$
which leads to
$$
\begin{split}
\hat{\mathbf{\theta}}_\text{MAP}(\mathbf{x}) 
&\stackrel{\text{def}}{=} \argmax_{\mathbf{\theta}} f_{\mathbf{\Theta}|\mathbf{X}}(\mathbf{\theta}|\mathbf{x})
\\
&= \argmax_{\mathbf{\theta}} \Bigl(
	\log f_{\mathbf{X}|\mathbf{\Theta}}(\mathbf{x}|\mathbf{\theta}) + 
	\log f_\mathbf{\Theta}(\mathbf{\theta})
\Bigr)
\\
&\equiv \argmax_{\mathbf{\theta}} \Bigl(
	\log f_\mathbf{X}(\mathbf{x}; \mathbf{\theta}) + 
	\log f_\mathbf{\Theta}(\mathbf{\theta})
\Bigr).
\end{split}
$$

> In the second step we dropped the constant $\frac{1}{f_\mathbf{X}(\mathbf{x})}$ during the expansion as it does not depend on $\mathbf{\theta}$.

Note how, in the next step, we considered the likelihood of the ML estimator instead of the MAP one. They are not equal, as the former is deterministic while the latter is not, but they give the same value. This trick is meant to allow an important insight: the $\hat{\mathbf{\theta}}_\text{MAP}(\mathbf{x})$ swings between $\hat{\mathbf{\theta}}_\text{ML}(\mathbf{x})$ and $f_\mathbf{\Theta}(\mathbf{\theta})$, depending on the situation, such as prior uncertainty of observation size.

| action                                     | effect                                                                 |
|--------------------------------------------|------------------------------------------------------------------------|
| $\uparrow\! N$ or $\uparrow\!\sigma_0$     | $\hat{\mathbf{\theta}}_\text{MAP} \to \hat{\mathbf{\theta}}_\text{ML}$ |
| $\downarrow\! N$ or $\downarrow\!\sigma_0$ | $\hat{\mathbf{\theta}}_\text{MAP} \to \mu_0$                           |

## Confidence Intervals

Now we proceed to quantify how good are those estimates. In other words, we are interested in the confidence of these estimates according to some statistical significance.

Let's begin looking at the randomness of an estimator $\hat{\mathbf{\Theta}}$. When we perform multiple finite samplings we get an estimate $\hat{\mathbf{\theta}}$ of the true parameter of interest, which can deviate from the true value. This distribution of the estimate is what generates the random variable $\hat{\mathbf{\Theta}}$. Therefore, when we report it, we must inform both its _value_ and _confidence_, as the latter is crucial for determining the quality of the reported estimate.

A **confidence interval** $\mathcal{I} = [\hat{\Theta} - \varepsilon, \hat{\Theta} + \varepsilon]$ is a statement about the estimator, and is itself random considering that new realizations could lead to different interval boundaries. Therefore, we are interested in answering whether the interval contains the true parameter, and we want to answer that with some **confidence level** of $1 - \alpha$. If we set $\alpha = 0.05$, for example, we say that $\mathcal{I}$ is a _95% confidence interval_.

$$ \mathbb{P}\bigl[ \theta \in \mathcal{I} \bigr] = \mathbb{P}\bigl[ \hat{\Theta} - \varepsilon \le \theta \le \hat{\Theta} + \varepsilon \bigr] \ge 1 - \alpha $$

> The confidence level _does not_ say how much of the population resides within the interval. Instead, it represents the chance that a random region may contain the true parameter.

### Building Intervals Knowing the Variance

Suppose we have $N$ i.i.d. observations $X_1, \dots, X_N$, that our estimator is the sample average (like the one obtained by the ML estimation) $\hat{\Theta}_\mu = \frac{1}{N} \sum_n X_n$, for which $\mathbb{E}[\hat{\Theta}_\mu] = \theta_\mu$ and $\text{Var}[\hat{\Theta}_\mu] = \frac{\sigma^2}{N}$, and that the population variance $\sigma^2$ is known but the mean $\mu$ is not (obviously, as we are using a mean estimator). As this estimator converges in distribution to a Gaussian distribution (by the CLT), we can construct a confidence interval by finding the $\varepsilon$ that covers $((1 - \alpha) \times 100)\%$ of the distribution generated by $\hat{\Theta}_\mu$, giving the following inequality to be solved:
$$\mathbb{P}\Bigl[ |\hat{Z}| \le \varepsilon \Bigr] \ge 1 - \alpha,$$
where
$$\hat{Z} \stackrel{\text{def}}{=} \frac{\hat{\Theta}_\mu - \mathbb{E}[\hat{\Theta}_\mu]}{\sqrt{ \text{Var}[\hat{\Theta}_\mu] }}$$
is a zero-mean unit-variance Gaussian (standardized variable). The solution is $\mathbb{P}[|\hat{Z}| \le \varepsilon] = 2\Phi(\varepsilon) - 1 \ge 1 - \alpha$, resulting in $\varepsilon \ge \Phi^{-1}\!\left( 1 - \frac{\alpha}{2} \right) = z_{\alpha/2}$ (critical value) and finally
$$
\mathcal{I} = \left[\;
	\hat{\Theta}_\mu - z_{\alpha/2} \frac{\sigma}{\sqrt{N}} 
	\quad,\quad
	\hat{\Theta}_\mu + z_{\alpha/2} \frac{\sigma}{\sqrt{N}} 
\;\right].
$$

> This definition considers that $\hat{\Theta}_\mu \sim \mathcal{N}(\theta_\mu, \frac{\sigma^2}{N})$. If it is not the case, we must replace the CDF when solving $\mathbb{P}[\, |\hat{Z}| \le \varepsilon \,]$.

> The standard deviation of the estimator is called **standard error**: $\text{SE} = \sqrt{\text{Var}[\hat{\Theta}_\mu]} = \frac{\sigma}{\sqrt{N}}$.

> The margin of error is defined as $z_{\alpha/2}\,\text{SE}$.

### Building Intervals Not Knowing the Variance

This definition assumes that the population variance is given a priori, which is not always the case. Sometimes we need to use an estimator for the variance as well. Let's say we have the mean estimator $\hat{\Theta}_\mu = \frac{1}{N} \sum_n X_n$ and the variance estimator $\hat{\Theta}_\sigma^2 = \frac{1}{N-1} \sum_n (X_n - \hat{\Theta}_\mu)^2$, both unbiased. In order to find a confidence interval $\mathcal{I}$ that contains the true mean $\mu$ we can repeat the same steps, but the resulting distribution is not a standard Gaussian anymore, but a **$t$-distribution** with $N-1$ degrees of freedom (which is very similar to the Gaussian and actually converges in distribution, but for really small $N$ it shows a heavy tail). Then, the resulting standardized random variable is
$$\hat{T} \stackrel{\text{def}}{=} \frac{\hat{\Theta}_\mu - \mu}{\hat{\Theta}_\sigma / \sqrt{N}}.$$

The $t$-distribution with $u$ degrees of freedom has the following PDF
$$
f_X(x) = \frac{
	\Gamma( \frac{u+1}{2} )
}{
	\sqrt{u\pi} \; \Gamma( \frac{u}{2} )
} \,
\left(
	1 + \frac{x^2}{u}
\right)^{
	- \,\frac{u+1}{2}
}
$$
and a CDF of $X$ at $z$ denoted by $\Psi_u(z)$ for what we compute $z_{\alpha/2} = \Psi_u^{-1}(1 - \frac{\alpha}{2})$ taking $u = N-1$. Finally, the confidence interval is
$$
\mathcal{I} = \left[\;
	\hat{\Theta}_\mu - z_{\alpha/2} \frac{\hat{\Theta}_\sigma}{\sqrt{N}} 
	\quad,\quad
	\hat{\Theta}_\mu + z_{\alpha/2} \frac{\hat{\Theta}_\sigma}{\sqrt{N}} 
\;\right].
$$

> In this case, the standard error is denoted by $\hat{\text{SE}} = \sqrt{\text{Var}[\hat{\Theta}_\mu]} = \frac{\hat{\Theta}_\sigma}{\sqrt{N}}$ and the margin of error is $z_{\alpha/2}\,\hat{\text{SE}}$.

> The estimate variance is not the same as standard error. The former approximates the dispersion around the true mean, while the latter quantifies the dispersion of the mean estimates.

> For $N \le 30$, the $t$-distribution is preferred over standard Gaussian.

### Building Intervals with Bootstrapping

One can ask "what if we want to estimate another statistic other than the true mean (e.g. median, mode...)?". Well, that can be a problem. First, not all statistics have simple expressions for the confidence interval, also the assumptions that allow us to apply the CLT may not hold. The **bootstrapping** method consists in a procedure to generate a set of synthetic datasets to mimic the unknown population based on the observed dataset, from where we can compute the statistics of interest and construct the confidence interval.

The intuition is that if we had access to the distribution of the population, we could extract multiple datasets, compute our desired statistic per dataset and measure its mean and variance in order to create the confidence interval. As we do not have access to the population and we have only one dataset, we treat the empirical distribution as the population, and as long as the former CDF approximates well the latter one the "plug-in principle" should work.

Consider a dataset $\mathcal{X} = \{x_1, \dots, x_N\}$ with empirical CDF $\hat{F}_X(x)$ (which assigns probability $1/N$ to each observed point $x_n$) and let $\hat{\Theta} = g(\mathcal{X})$ be an estimator of an arbitrary statistic $g$. We can extract $K$ samples of size $N$ with replacement from $\mathcal{X}$ (called _bootstrap replicate datasets_):
$$
\begin{split}
\mathcal{Y}^{(1)} &= \{Y_1^{(1)}, \dots, Y_N^{(1)}\} \sim \hat{F}_X,
\\
\vdots
\\
\mathcal{Y}^{(K)} &= \{Y_1^{(K)}, \dots, Y_N^{(K)}\} \sim \hat{F}_X.
\end{split}
$$

For each replicate dataset, we compute the statistic $\hat{\Theta}_\text{boot}^{(k)} = g\!\left( \mathcal{Y}^{(k)} \right)$ for $k = 1, \dots, K$ and then the sample mean and variance
$$
\begin{split}
\hat{\mu}_\text{boot} &\stackrel{\text{def}}{=} \frac{1}{K} \sum_{k = 1}^K \hat{\Theta}_\text{boot}^{(k)},
\\
\hat{\sigma}_\text{boot}^2 &\stackrel{\text{def}}{=} \frac{1}{K} \sum_{k = 1}^K (\hat{\Theta}_\text{boot}^{(k)} - \hat{\mu}_\text{boot})^2.
\end{split}
$$

Defining the bootstrap standard error as $\hat{\text{SE}}_\text{boot} \stackrel{\text{def}}{=} \sqrt{\hat{\sigma}_\text{boot}^2}$, we can construct a normal-approximation confidence interval as
$$
\mathcal{I} = \left[\;
	\hat{\Theta} - z_{\alpha/2} \,\hat{\text{SE}}_\text{boot}
	\quad,\quad
	\hat{\Theta} + z_{\alpha/2} \,\hat{\text{SE}}_\text{boot}
\;\right].
$$

This method relies on two layers of approximation:
- _Monte Carlo approximation_ ($K \to \infty$): By the law of large numbers, as the number of bootstrap replicates $K \to \infty$, $\hat{\mu}_\text{boot}$ and $\hat{\sigma}_\text{boot}^2$ converge to the exact expectation and variance of $g(\mathcal{Y})$ under the empirical distribution $\hat{F}_X$.
- _Plug-in approximation_ ($N \to \infty$): As the original sample size $N \to \infty$, the empirical distribution $\hat{F}_X$ converges to the true population distribution $F_X$, so the bootstrap sampling distribution approximates the true sampling distribution of $\hat{\Theta}$.

> The validity of the interval will follow the validity of $z_{\alpha/2}$. If $\hat{\Theta}$ is roughly a Gaussian (the usual case) this approach is fine, otherwise we must use the empirical quantiles at $\frac{\alpha}{2}$ and $1 - \frac{\alpha}{2}$.

## Hypothesis Testing

We can understand an hypothesis as a statement about the data. As observations change, the conclusion can change accordingly. Therefore, an **hypothesis testing** is a formal way of taking decisions based on the available evidence and a cutoff threshold. Usually we have two hypothesis:
- The _null hypothesis_ $H_0$, representing the "status quo" or the current situation;
- The _alternative hypothesis $H_1$, representing a different proposal from the null one.

The null hypothesis is the default choice, meaning that in the lack of enough evidence we must stick with it. Notice that it does not mean the alternative hypothesis is wrong, just that we cannot reach the minimum necessary confidence to choose it. In fact, is impossible to say if either one is correct or wrong, as the truth is unknown. Because of that we care about making the best possible decision, given the provided information.
